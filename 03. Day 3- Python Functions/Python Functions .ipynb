-----------Python Functions-----------
A function is a block of code that performs a specific task whenever it is called. There are two types of functions: built-in functions user-defined functions

built-in functions: These functions are defined and pre-coded in python. examples: min(), max(), len(), sum(), type(), range(), dict(), list(), tuple(), set(), print(), etc

user-defined functions: We can create functions to perform specific tasks as per our needs. Such functions are called user-defined functions.

```python
def name(fname, lname):
    print("Hello,", fname, lname)

name("Jack", "David")
```
